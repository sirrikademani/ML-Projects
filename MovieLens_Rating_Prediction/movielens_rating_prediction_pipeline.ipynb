The problem — power law distributions
Your ratings per user look like this:
Most users:  8 ratings
Some users:  100 ratings  
Few users:   3,381 ratings
If you plot it, it's a hockey stick — a tiny number of users have extreme values. This is called a power law or long tail.

Why this hurts ML models
Most ML algorithms assume features are roughly on a similar scale. When one user has 8 ratings and another has 3,381, that gap of 3,373 dominates the model. The algorithm pays huge attention to that difference and less to everything else. This causes:

Poor generalisation
Feature importance being distorted
Coefficients being unstable


What log transform does
np.log1p(x) = log(x + 1) — the +1 handles zeros safely.
It compresses large values while preserving the order:
Raw count  →  log1p
1          →  0.69
8          →  2.20
17         →  2.94
100        →  4.62
1,000      →  6.91
3,381      →  8.13
The gap between 1 and 3,381 goes from 3,380 → 7.44. The order is preserved — a user with more ratings still has a higher value — but the extreme values no longer dominate.

Why +1 specifically?
Because log(0) is undefined (negative infinity). Adding 1 ensures the minimum input is 1, giving log(1) = 0 cleanly. That's why it's log1p not just log.

Rule of thumb: whenever you see count data or monetary values with a long tail — ratings, revenue, page views, salaries — log transform is almost always the right move before feeding into a model.
Ready to continue with the output?

### Hypothesis Testing

<b>The hypothesis Test 1:</b>


1990 is a natural dividing line — pre-1990 films are classic/vintage, post-1990 are modern. The hypothesis is whether nostalgia or classic status inflates ratings for older films. Could have used 1980 or 2000 — the exact cutoff is a judgement call you should justify in your write-up.

H₀: Old movies (pre-1990) and new movies (post-1990) are rated the same <br>
H₁: There is a significant difference in ratings between old and new movies

```python
old = df_model[df_model['year']<1990]['rating']
new = df_model[df_model['year']>=1990]['rating']
```

<b>Run Welch's t-test on the two groups.</b>

Returns:

> t1 — how many standard errors apart the two means are <br>
> p1 — probability of seeing this difference by chance if H₀ were true

t1: t-statistic:"How many standard errors apart are the two group means?" <br>
t = (mean_old - mean_new) / standard_error <br>

t = 0          → means are identical, no difference at all<br>
t = 1 or -1    → means are 1 standard error apart, weak signal<br>
t = 2 or -2    → getting significant (~p=0.05)<br>
t = 3+         → strong signal, almost certainly significant<br>
t = 50+        → what you'll likely see with 2.9M rows<br>

t  →  tells you direction + how strong the signal is relative to noise<br>
p  →  tells you if it's statistically significant<br>
mean difference  →  tells you if it's practically meaningful<br>

ttest_ind = t-test for two independent groups. Here's what it does under the hood:
pythont_stat, p_value = stats.ttest_ind(group_a, group_b, equal_var=False)
Step 1 — Calculate means and variances of both groups:
mean_a, var_a, n_a = old.mean(), old.var(), len(old)
mean_b, var_b, n_b = new.mean(), new.var(), len(new)
Step 2 — Calculate standard error (Welch's formula):
SE = sqrt(var_a/n_a + var_b/n_b)
Step 3 — Calculate t-statistic:
t = (mean_a - mean_b) / SE
This is just "how many standard errors apart are the two means?"
Step 4 — Calculate degrees of freedom (Welch-Satterthwaite equation):
df = (var_a/n_a + var_b/n_b)²  /  ((var_a/n_a)²/(n_a-1) + (var_b/n_b)²/(n_b-1))
This adjusts df downward when variances are unequal — making the test more conservative.
Step 5 — Get p-value from t-distribution:
p = 2 * (1 - CDF(|t|, df))   ← two-tailed

```python
# perform t test of 2 independent groups old and new [not paired, seperate groups]
#equal_var=True   →  Student's t-test (assumes same variance in both groups)
#equal_var=False  →  Welch's t-test (no assumption about variance) ← safer

t1, p1 = stats.ttest_ind(old,new, equal_var=False)
```

Always print the means first — the t-test tells you if there's a difference, the means tell you which direction and how big.

```python
print(f'Old mean: {old.mean():.3f}, New mean: {new.mean():.3f}')
print(f"t={t1:.4f}, p={p1:.6f}")
print("→ Reject H0" if p1 < 0.05 else "→ Fail to reject H0")
print()
```

<b>The hypothesis Test 2:</b>


100+ = power users who rate systematically and critically. ≤10 = casual users who only bother rating movies they feel strongly about (usually loved or hated). The hypothesis is whether these groups rate differently. Note we deliberately skipped the middle (11–100) to get a clean contrast between extremes.

H₀: > 100 <br>
Ha: <= 10 ratings

```python
# Test 2 — Prolific vs casual users
prolific = df_model[df_model['rating_count_per_user'] > 100]['rating']
casual = df_model[df_model['rating_count_per_user'] <= 10]['rating']
t2, p2 = stats.ttest_ind(prolific, casual, equal_var=False)
print(f"Test 2 — Prolific vs casual raters")
print(f"Prolific mean: {prolific.mean():.3f}, Casual mean: {casual.mean():.3f}")
print(f"t={t2:.4f}, p={p2:.6f}")
print("→ Reject H0" if p2 < 0.05 else "→ Fail to reject H0")

print()
```

<b>Test 3 — Action vs Romance </b><br>
Ho: pythonaction  <br>
Ha: romance <br>
These genres attract very different audiences with different rating behaviours. Action fans tend to rate more generously, romance watchers more critically — or vice versa. The genre dummy columns make this trivial to split.

```python
# Test 3 — Action vs Romance ratings
action = df_model[df_model['Action'] == 1]['rating']
romance = df_model[df_model['Romance'] == 1]['rating']
t3, p3 = stats.ttest_ind(action, romance, equal_var=False)
print(f"Test 3 — Action vs Romance")
print(f"Action mean: {action.mean():.3f}, Romance mean: {romance.mean():.3f}")
print(f"t={t3:.4f}, p={p3:.6f}")
print("→ Reject H0" if p3 < 0.05 else "→ Fail to reject H0")

print()

```

<b>Test 4 — avg_rating >= 4.0 vs < 3.0</b><br>
Ho: high_movie <br>
Ha: low_movie  <br>
This tests whether movie quality (as measured by crowd average) actually predicts individual ratings — which it should, but by how much? Again skipping the middle (3.0–4.0) to get a clean signal.

```python
# Test 4 — High vs low rated movies affect on user rating
high_movie = df_model[df_model['avg_rating_per_movies'] >= 4.0]['rating']
low_movie  = df_model[df_model['avg_rating_per_movies'] < 3.0]['rating']
t4, p4 = stats.ttest_ind(high_movie, low_movie, equal_var=False)
print(f"Test 4 — High quality vs low quality movies")
print(f"High mean: {high_movie.mean():.3f}, Low mean: {low_movie.mean():.3f}")
print(f"t={t4:.4f}, p={p4:.6f}")
print("→ Reject H0" if p4 < 0.05 else "→ Fail to reject H0")
```

### Phase4 - Modeling

```python
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import (mean_squared_error, r2_score, 
                             classification_report, roc_auc_score,
                             confusion_matrix)
from sklearn.preprocessing import StandardScaler
```

```python
# define features
genre_cols = ['Action','Adventure','Animation','Children','Comedy','Crime',
              'Documentary','Drama','Fantasy','Film-Noir','Horror','IMAX',
              'Musical','Mystery','Romance','Sci-Fi','Thriller','War','Western']
```

```python
feature_cols = [
    'avg_rating_per_movies',   # movie quality signal
    'rating_count_per_movie', # movie popularity
    'avg_rating_per_user',    # user bias
    'rating_count_per_user',  # user activity
    'log_movie_count',        # log-transformed popularity
    'log_user_count',         # log-transformed activity
    'movie_age',              # how old is the movie
] + genre_cols
```

```python
# drop rows with any nulls in features
df_clean = df_model[feature_cols + ['rating', 'liked']].dropna()
print(f"Modelling dataset: {df_clean.shape}")
print(f"Nulls: {df_clean.isnull().sum().sum()}")
```

```python
# features and targets
X = df_clean[feature_cols]
y_reg = df_clean['rating']      # regression target
y_clf = df_clean['liked']       # classification target

# train/test split — stratify on liked to preserve 63/37 class balance
X_train, X_test, y_train_reg, y_test_reg, y_train_clf, y_test_clf = train_test_split(
    X, y_reg, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)

print(f"Train: {X_train.shape[0]:,} rows")
print(f"Test:  {X_test.shape[0]:,} rows")
print(f"Train liked %: {y_train_clf.mean()*100:.1f}%")
print(f"Test liked %:  {y_test_clf.mean()*100:.1f}%")

# scale features — important for logistic regression
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
```

```python
# ── REGRESSION ──────────────────────────────────────────
reg = LinearRegression()
reg.fit(X_train, y_train_reg)
y_pred_reg = reg.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg))
r2   = r2_score(y_test_reg, y_pred_reg)

print("=" * 40)
print("REGRESSION — Predict rating (1–5)")
print("=" * 40)
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
print(f"Baseline RMSE (always predict mean): {np.sqrt(mean_squared_error(y_test_reg, [y_train_reg.mean()]*len(y_test_reg))):.4f}")

# feature importance
coef_df = pd.DataFrame({
    'feature': feature_cols,
    'coefficient': reg.coef_
}).sort_values('coefficient', ascending=False)
print("\nTop 5 features:")
print(coef_df.head())
print("\nBottom 5 features:")
print(coef_df.tail())
```

```python
# ── CLASSIFICATION ──────────────────────────────────────
clf = LogisticRegression(max_iter=1000, random_state=42)
clf.fit(X_train_scaled, y_train_clf)
y_pred_clf  = clf.predict(X_test_scaled)
y_pred_prob = clf.predict_proba(X_test_scaled)[:, 1]

print("=" * 40)
print("CLASSIFICATION — Predict liked (0/1)")
print("=" * 40)
print(classification_report(y_test_clf, y_pred_clf, 
                             target_names=['Not liked', 'Liked']))
print(f"AUC-ROC: {roc_auc_score(y_test_clf, y_pred_prob):.4f}")
print(f"Baseline accuracy (always predict liked): {y_test_clf.mean()*100:.1f}%")
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_clf, y_pred_clf))
```

### Conclusion
* Liked recall = 87% — model catches 87% of movies the user will like. Great for recommendations — you don't want to miss movies someone would love.
* Not liked recall = 55% — model misses 45% of movies the user won't like. This is your weakness — it over-recommends.
* This is expected given class imbalance (63% liked) — model is biased toward predicting liked.
* RMSE improved 20% over baseline — your model predicts within ±0.84 stars on average vs the dumb baseline of always predicting the mean (±1.05 stars). Not bad for a linear model with no collaborative filtering.
* R² of 0.36 means your features explain 36% of the variance in ratings. The remaining 64% is driven by things you don't have — personal taste, mood, context. This is actually expected and honest for rating prediction.


#### Strengths:

* 20% RMSE improvement over baseline
* AUC-ROC 0.80 with a simple linear baseline model
* Confirmed hypothesis testing findings — user bias and movie quality dominate
* Stratified split preserved class balance perfectly
